## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Binary Classificaiton because
Instead of just applying static rules, we frame this as a binary classification problem to predict whether a specific page is experiencing true performance decay (is_decaying = 1 when trend_direction == 'down')

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target / Proxy: is_decaying (Binary target: 1 for decaying pages, 0 for healthy/stable pages)

Source & Logic:
The target label is directly derived from the observed historical outcome column trend_direction in our dataset.

Specifically, we map trend_direction == 'down' to 1 (indicating true organic performance loss) and all other trend states (stable, up, flat, new) to 0. Since we cannot directly observe future organic traffic drop before it happens, this historical decay indicator serves as our proxy target to train the model to detect pages at high risk of performance loss.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Success Metric: ROC-AUC and Precision@K (e.g., Precision@Top-100)

Defensible Threshold for 'Good':
ROC-AUC >0.80: Evaluates the model's overall capability to discriminate between decaying pages (is_decaying = 1) and healthy pages across all potential classification thresholds. It directly validates the model's ranking quality regardless of class imbalance.
Precision@Top-100 >0.85: Ensures that at least 85% of the top 100 highest-risk URLs prioritized for an editorial update are true positive decay cases. This directly minimizes False Positives, preventing the team from wasting editorial budget and effort on healthy content.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

Unit of Analysis: One Row = One Unique Content URL / Asset (content_id)

Each observation in our dataset represents the search metrics, metadata, and historical performance trend of a single content URL. Our model operates at this single-asset resolution to issue granular page-level refresh recommendations.

```python
import pandas as pd
import os

file_path = "content_refresh_anonymized.csv"

df = pd.read_csv(file_path)

df['is_decaying'] = (df['trend_direction'].str.lower() == 'down').astype(int)

print(f"Dataframe Boyutu: {df.shape[0]} satır, {df.shape[1]} sütun")
print("Analiz Birimi: 1 Satır = 1 Özgün İçerik URL'si (content_id)\n")

cols_to_show = ['content_id', 'search_volume', 'days_since_last_update', 'trend_direction', 'is_decaying']
df[cols_to_show].head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why Static Rules Fail:
A simple heuristic rule (such as if days_since_last_update > 180) achieves only a 47.1% decay detection rate, as demonstrated in our Week 1 exploratory analysis.

Organic performance decay is rarely driven by a single attribute. It is governed by complex, non-linear feature interactions between search volume (search_volume), competitive intensity (competition), content metadata (content_type, word_count), and update frequency. For example, a 90-day-old page in a high-competition category may decay rapidly, while a 300-day-old page targeting a low-competition niche remains stable.

Why ML Beats Hardcoded Logic:
Fixed conditional statements (if-else) cannot adaptively scale or capture multi-variable decay patterns across tens of thousands of URLs simultaneously. A machine learning classification model captures these non-linear feature interactions and subtle trend shifts, accurately isolating true declining assets from statistical noise while minimizing wasted editorial budget on healthy pages.
